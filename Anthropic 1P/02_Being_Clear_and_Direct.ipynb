## Cours

**Claude répond le mieux à des instructions claires et directes.**

Voyez Claude comme une personne qui débute dans un poste. **Claude n'a aucun contexte** sur ce qu'il doit faire, en dehors de ce que vous lui dites littéralement. Comme lorsque vous expliquez une tâche à quelqu'un pour la première fois, plus vous exposez simplement et précisément ce que vous voulez, meilleure et plus exacte sera la réponse de Claude.

En cas de doute, appliquez la **règle d'or du prompt clair** :
- Montrez votre prompt à un collègue ou à un ami et demandez-lui de suivre lui-même les instructions, pour voir s'il obtient le résultat attendu. S'il est perdu, Claude l'est aussi.

### Exemples

Prenons une tâche comme l'écriture de poésie. (Ignorez les erreurs de nombre de syllabes : les LLM ne savent pas encore bien compter les syllabes.)

```python
# Prompt
PROMPT = "Write a haiku about robots."

# Afficher la réponse de Claude
print(get_completion(PROMPT))
```

Ce haïku est correct, mais l'utilisateur peut vouloir que Claude commence directement par le poème, sans le préambule « Here is a haiku ».

Comment l'obtenir ? Il suffit de **le demander** !

```python
# Prompt
PROMPT = "Write a haiku about robots. Skip the preamble; go straight into the poem."

# Afficher la réponse de Claude
print(get_completion(PROMPT))
```

Autre exemple : demandons à Claude qui est le meilleur joueur de basket de tous les temps. Comme vous le voyez ci-dessous, Claude cite plusieurs noms mais **ne désigne pas un « meilleur » unique**.

```python
# Prompt
PROMPT = "Who is the best basketball player of all time?"

# Afficher la réponse de Claude
print(get_completion(PROMPT))
```

Peut-on amener Claude à trancher et à choisir un meilleur joueur ? Oui, il suffit de le demander !

```python
# Prompt
PROMPT = "Who is the best basketball player of all time? Yes, there are differing opinions, but if you absolutely had to pick one player, who would it be?"

# Afficher la réponse de Claude
print(get_completion(PROMPT))
```

### Exercice 2.1 - Espagnol
Modifiez le `SYSTEM_PROMPT` pour que Claude réponde en espagnol.

```python
# Prompt système : c'est le seul champ à modifier
SYSTEM_PROMPT = "[Replace this text]"

# Prompt
PROMPT = "Hello Claude, how are you?"

# Obtenir la réponse de Claude
response = get_completion(PROMPT, SYSTEM_PROMPT)

# Fonction de correction de l'exercice
def grade_exercise(text):
    return "hola" in text.lower()

# Afficher la réponse de Claude et le résultat de la correction
print(response)
print("\n--------------------------- GRADING ---------------------------")
print("This exercise has been correctly solved:", grade_exercise(response))
```

### Exercice 2.2 - Un seul joueur

Modifiez le `PROMPT` pour que Claude ne tergiverse pas du tout et réponde **UNIQUEMENT** par le nom d'un joueur précis, **sans aucun autre mot ni ponctuation**.

```python
# Prompt : c'est le seul champ à modifier
PROMPT = "[Replace this text]"

# Obtenir la réponse de Claude
response = get_completion(PROMPT)

# Fonction de correction de l'exercice
def grade_exercise(text):
    return text == "Michael Jordan"

# Afficher la réponse de Claude et le résultat de la correction
print(response)
print("\n--------------------------- GRADING ---------------------------")
print("This exercise has been correctly solved:", grade_exercise(response))
```

### Exercice 2.3 - Écrire une histoire

Modifiez le `PROMPT` pour que Claude produise la réponse la plus longue possible. Si la réponse dépasse **800 mots**, l'exercice est validé.

```python
# Prompt : c'est le seul champ à modifier
PROMPT = "[Replace this text]"

# Obtenir la réponse de Claude
response = get_completion(PROMPT)

# Fonction de correction de l'exercice
def grade_exercise(text):
    trimmed = text.strip()
    words = len(trimmed.split())
    return words >= 800

# Afficher la réponse de Claude et le résultat de la correction
print(response)
print("\n--------------------------- GRADING ---------------------------")
print("This exercise has been correctly solved:", grade_exercise(response))
```
