## Cours

Dans la continuité de l'idée que Claude n'a pas d'autre contexte que ce que vous lui dites, il est parfois important de **lui demander d'endosser un rôle précis (avec tout le contexte nécessaire)**. C'est ce qu'on appelle le role prompting. Plus le rôle est décrit en détail, mieux c'est.

**Donner un rôle à Claude peut améliorer ses performances** dans de nombreux domaines, de la rédaction au code en passant par la synthèse. C'est comparable à l'aide qu'apporte à une personne la consigne « réfléchis comme un ______ ». Le role prompting peut aussi modifier le style, le ton et la manière de répondre.

**Note :** le rôle peut être donné dans le prompt système ou dans le message du tour User.

### Exemples

Dans l'exemple ci-dessous, sans rôle attribué, Claude donne une **réponse directe et sans style particulier** lorsqu'on lui demande son point de vue en une phrase sur le skateboard.

En revanche, lorsqu'on lui demande d'endosser le rôle d'un chat, son point de vue change : **le ton, le style et le contenu de sa réponse s'adaptent au nouveau rôle**.

**Note :** une technique complémentaire consiste à **indiquer à Claude le public auquel il s'adresse**. Ci-dessous, on aurait pu préciser à qui Claude doit parler. « Tu es un chat » ne donne pas du tout la même réponse que « tu es un chat qui parle à une foule de skateurs ».

Voici le prompt sans rôle dans le prompt système :

```python
# Prompt
PROMPT = "In one sentence, what do you think about skateboarding?"

# Afficher la réponse de Claude
print(get_completion(PROMPT))
```

Voici la même question, cette fois avec un rôle attribué.

```python
# Prompt système
SYSTEM_PROMPT = "You are a cat."

# Prompt
PROMPT = "In one sentence, what do you think about skateboarding?"

# Afficher la réponse de Claude
print(get_completion(PROMPT, SYSTEM_PROMPT))
```

Le role prompting permet de faire imiter à Claude certains styles d'écriture, de lui faire adopter une certaine voix, ou de régler la complexité de ses réponses. **Il peut aussi rendre Claude plus performant en mathématiques et en logique.**

Dans l'exemple ci-dessous, il existe une réponse correcte et sans ambiguïté : oui. Pourtant Claude se trompe et pense manquer d'informations, alors que ce n'est pas le cas :

```python
# Prompt
PROMPT = "Jack is looking at Anne. Anne is looking at George. Jack is married, George is not, and we don’t know if Anne is married. Is a married person looking at an unmarried person?"

# Afficher la réponse de Claude
print(get_completion(PROMPT))
```

Et si on **demandait à Claude de se comporter comme un robot logicien** ? Comment sa réponse change-t-elle ?

Avec ce nouveau rôle, Claude trouve la bonne réponse. (Pas forcément pour toutes les bonnes raisons, cela dit.)

```python
# Prompt système
SYSTEM_PROMPT = "You are a logic bot designed to answer complex logic problems."

# Prompt
PROMPT = "Jack is looking at Anne. Anne is looking at George. Jack is married, George is not, and we don’t know if Anne is married. Is a married person looking at an unmarried person?"

# Afficher la réponse de Claude
print(get_completion(PROMPT, SYSTEM_PROMPT))
```

**Note :** vous verrez tout au long de ce cours qu'il existe **de nombreuses techniques de prompt engineering pour obtenir des résultats similaires**. Le choix vous appartient ! Nous vous encourageons à **expérimenter pour trouver votre propre style**.

### Exercice 3.1 - Correction de calcul
Dans certains cas, **Claude peut avoir du mal avec les mathématiques**, même simples. Ci-dessous, Claude juge correcte la résolution d'un problème alors qu'il y a une erreur de calcul évidente à la deuxième étape. Notez que Claude repère bien l'erreur lorsqu'il procède étape par étape, mais n'en conclut pas que la solution globale est fausse.

Modifiez le `PROMPT` et/ou le `SYSTEM_PROMPT` pour que Claude juge la solution `incorrectly` résolue (incorrecte) plutôt que correcte.



```python
# Prompt système : si vous ne voulez pas de prompt système, laissez cette variable à chaîne vide
SYSTEM_PROMPT = ""

# Prompt
PROMPT = """Is this equation solved correctly below?

2x - 3 = 9
2x = 6
x = 3"""

# Obtenir la réponse de Claude
response = get_completion(PROMPT, SYSTEM_PROMPT)

# Fonction de correction de l'exercice
def grade_exercise(text):
    if "incorrect" in text or "not correct" in text.lower():
        return True
    else:
        return False

# Afficher la réponse de Claude et le résultat de la correction
print(response)
print("\n--------------------------- GRADING ---------------------------")
print("This exercise has been correctly solved:", grade_exercise(response))
```
