# Annexe 10.3 : recherche et récupération d'informations

Saviez-vous que Claude peut **faire des recherches dans Wikipédia pour vous** ? Claude peut trouver et récupérer des articles, puis les résumer, en faire la synthèse, rédiger du contenu nouveau à partir de ce qu'il a trouvé, et bien plus. Et pas seulement sur Wikipédia : vous pouvez aussi faire des recherches dans vos propres documents, qu'ils soient stockés en texte brut ou indexés dans une base vectorielle.

Consultez nos [exemples RAG du cookbook](https://github.com/anthropics/claude-cookbooks/blob/main/third_party/Wikipedia/wikipedia-search-cookbook.ipynb) pour apprendre à compléter les connaissances de Claude et à améliorer la précision et la pertinence de ses réponses avec des données issues de bases vectorielles, de Wikipédia, d'internet, etc. Vous y verrez aussi comment utiliser certains outils d'[embeddings](https://docs.anthropic.com/claude/docs/embeddings) et de bases vectorielles.

Pour découvrir des architectures RAG avancées avec Claude, consultez nos [diapositives de présentation technique de Claude 3 sur les architectures RAG](https://docs.google.com/presentation/d/1zxkSI7lLUBrZycA-_znwqu8DDyVhHLkQGScvzaZrUns/edit#slide=id.g2c736259dac_63_782) (en anglais).
