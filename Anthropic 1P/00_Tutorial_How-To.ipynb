### Le SDK Anthropic et l'API Messages
Tout au long de ce tutoriel, nous utilisons le [SDK Python d'Anthropic](https://docs.anthropic.com/claude/reference/client-sdks) et l'[API Messages](https://docs.anthropic.com/claude/reference/messages_post).

Voici un exemple de l'exécution d'un prompt dans ce tutoriel. Nous commençons par créer `get_completion`, une fonction utilitaire qui envoie un prompt à Claude et renvoie la réponse générée. Exécutez cette cellule maintenant.

```python
import anthropic

client = anthropic.Anthropic(api_key=API_KEY)

def get_completion(prompt: str):
    message = client.messages.create(
        model=MODEL_NAME,
        max_tokens=2000,
        temperature=0.0,
        messages=[
          {"role": "user", "content": prompt}
        ]
    )
    return message.content[0].text
```

Nous allons maintenant écrire un exemple de prompt pour Claude et afficher sa réponse en appelant la fonction utilitaire `get_completion`. L'exécution de la cellule ci-dessous affiche la réponse de Claude juste en dessous.

N'hésitez pas à modifier la chaîne du prompt pour obtenir des réponses différentes.

```python
# Prompt
prompt = "Hello, Claude!"

# Obtenir la réponse de Claude
print(get_completion(prompt))
```
